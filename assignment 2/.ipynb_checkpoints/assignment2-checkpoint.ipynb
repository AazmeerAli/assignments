### 1. Write a program that checks if a given number is positive, negative or zero.

```python
number = 0

if number > 0:
    print("number is positive")
elif number < 0:
    print("number is negative")
else:
    print("number is zero")
```

### 2. Write a program to find the larger of the two numbers. 

```python
number_1 = 6
number_2 = 13

if number_1 > number_2:
    print(f"number 1 is larger: {number_1}")
else:
    print(f"number 2 is larger: {number_2}")
```

### 3. Write a program to find the largest of the three numbers. 

```python
number_1 = 1116
number_2 = 123
number_3 = 26

if number_1 > number_2 and number_1 > number_3:
    print(f"number 1 is larger: {number_1}")
elif number_2 > number_1 and number_2 > number_3:
    print(f"number 2 is larger: {number_2}")
else:
    print(f"number 3 is larger: {number_3}")
```

### 4. Write a program to check if the given string “Mass” is present in the string “Saylani Mass IT”. If the string is present then display the message “string found”.

### 5. Write a program to take user’s age as input and display whether he is minor, adult or senior citizen:
###### a. Minor age is less than 18.
###### b. Adult age is greater than 18 and less than 60
###### c. Senior citizen age is greater than 60

```python
age = int(input("Enter your age: "))

if age < 18:
    print("You are a minor.")
elif age >= 18 and age < 60:
    print("You are an adult.")
else:
    print("You are a senior citizen.")
```

### 6. Write a program to take integer as input and check if it is even or odd. 

```python
number = int(input("Enter an integer: "))

if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

### 7. Write a program to take two numbers and an operator (/,x,+,-) as input and perform the corresponding operation. 

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
operator = input("Enter operator (+, -, x, /): ")

if operator == "+":
    print(num1 + num2)
elif operator == "-":
    print(num1 - num2)
elif operator == "x":
    print(num1 * num2)
elif operator == "/":
    print(num1 / num2)
else:
    print("Invalid operator")
```

### 8. Write a program to take number as input and check if it lies in the range i.e. 20 (inclusive) – 40 (exclusive) 

```python
number = int(input("Enter a number: "))

if number >= 20 and number < 40:
    print("Number lies in the range.")
else:
    print("Number does not lie in the range.")
```

### 9. Write a program that ask for an integer number and checks if it is divisible by 2, 3, or both. 

```python
number = int(input("Enter an integer: "))

if number % 2 == 0 and number % 3 == 0:
    print("The number is divisible by both 2 and 3.")
elif number % 2 == 0:
    print("The number is divisible by 2.")
elif number % 3 == 0:
    print("The number is divisible by 3.")
else:
    print("The number is not divisible by 2 or 3.")
```

### 10. Write a program to take user’s score as input and determine if they pass or fail (pass if score is above 60, otherwise fails.)

### 11. Write a program that takes a temperature in Celsius and checks if it is freezing, moderate or hot.
###### a. Freezing temperature is below 0.
###### b. Moderate temperature is greater than 0 and less than 26.
###### c. Hot temperature is above 26.

```python
temperature = int(input("Enter temperature in Celsius: "))

if temperature <= 0:
    print("Freezing")
elif temperature > 0 and temperature <= 26:
    print("Moderate")
else:
    print("Hot")
```

### 12. Write a program to input salary, if salary is greater than Rs. 600000 then deduct 5% tax and calculate the Gross Salary. 

```python
salary = int(input("Enter your salary: "))

if salary > 600000:
    tax = salary * 0.05
    gross_salary = salary - tax
    print("Gross Salary:", gross_salary)
else:
    print("Gross Salary:", salary)
```

### 13. Write a program to input salary and deduct Tax and calculate gross salary
###### a. if salary is greater than 2500000 ---> 15% Tax Deduction
###### b. if salary is between 1500000 - 2500000 ---> 10% Tax Deduction
###### c. if salary is between 500000 - 1500000 ---> 5% Tax Deduction
###### d. if salary is less than 500000 ---> 0% Tax Deduction

```python
salary = int(input("Enter your salary: "))

if salary > 2500000:
    tax = salary * 0.15
elif salary >= 1500000:
    tax = salary * 0.1
elif salary >= 500000:
    tax = salary * 0.05
else:
    tax = 0

gross_salary = salary - tax

print("Tax:", tax)
print("Gross Salary:", gross_salary)
```
