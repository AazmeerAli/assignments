# Assignment 04: Library Management System

```python
books = []
loans = []
    
def find_book(book_id):
    for book in books:
        if book["id"] == book_id:
            return book
    return None
```

```python
def add_book():
    book_id = int(input("Book ID: "))

    if find_book(book_id) != None:
        print("This book ID is already exists!")
        return

    title = input("Title: ")
    author = input("Author: ")
    copies = int(input("Number of copies: "))

    book = {"id": book_id, "title": title, "author": author,
            "total": copies, "available": copies}
    books.append(book)
    print("Book added!")
```

```python
def view_books():
    if len(books) == 0:
        print("No books available.")
    else:
        for book in books:
            print(book["id"], "|", book["title"], "|", book["author"],
                  "| Total:", book["total"], "| Available:", book["available"])
```

```python
def search_book():
    book_id = int(input("Book ID: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
    else:
        print(book["id"], "|", book["title"], "|", book["author"],
              "| Total:", book["total"], "| Available:", book["available"])
```

```python
def update_book():
    book_id = int(input("Book ID: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
        return

    book["title"] = input("New title: ")
    book["author"] = input("New author: ")
    new_total = int(input("New total copies: "))

    difference = new_total - book["total"]
    book["total"] = new_total
    book["available"] = book["available"] + difference

    print("Book updated!")
```

```python
def delete_book():
    book_id = int(input("Book ID: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
    else:
        books.remove(book)
        print("Book deleted!")
```

```python
def issue_book():
    book_id = int(input("Book ID: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
    elif book["available"] == 0:
        print("Book is not available.")
    else:
        book["available"] = book["available"] - 1
        loans.append(book_id)
        print("Book issued!")
```

```python
def return_book():
    book_id = int(input("Book ID: "))
    book = find_book(book_id)

    if book == None:
        print("Book not found.")
    elif book_id not in loans:
        print("This book is not issued!")
    else:
        loans.remove(book_id)
        book["available"] = book["available"] + 1
        print("Book returned!")
```

```python
while True:
    print("\n1. Add Book")
    print("2. View All Books")
    print("3. Search Book")
    print("4. Update Book")
    print("5. Delete Book")
    print("6. Issue Book")
    print("7. Return Book")
    print("8. Exit")

    choice = input("Choose option: ")

    if choice == "1":
        add_book()
    elif choice == "2":
        view_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        update_book()
    elif choice == "5":
        delete_book()
    elif choice == "6":
        issue_book()
    elif choice == "7":
        return_book()
    elif choice == "8":
        print("Goodbye!")
        break
    else:
        print("Invalid option!")
```
