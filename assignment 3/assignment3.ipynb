### 12. You are organizing a party and need to enter the names of 10 guests. The program should check whether a guest's name has already been entered. Write a Python program that uses a for loop, list, input(), and if-else to manage a guest list and prevent duplicate names. 

```python
guests = []

for i in range(10):
    name = input("Enter guest name: ")

    while name in guests:
        print("Guest already exists! Enter a different name.")
        name = input("Enter guest name: ")

    guests.append(name)

print("Guest List:", guests)
```

### 13. You are creating a restaurant menu. The program should allow the user to enter 5 dish names. However, the restaurant does not serve fish, so the program should skip it whenever the user enters "fish". Write a Python program that uses a for loop, if condition, continue, and a list to create a restaurant menu while excluding a specific dish. 


```python
menu = []

for i in range(5):
    dish = input("Enter dish name: ")

    if dish == "fish":
        print("We don't serve fish")
        continue

    menu.append(dish)

print("Restaurant Menu:", menu)
```

### 14. Aap ek dawat ke liye menu aur uski cost calculate karne wala Python program banayein. Dawat mein 300 mehman hain. User se 5 dishes ke naam lene hain. Lekin kuch dishes menu mein allowed nahi hain. 


```python
discard_menu = ["fish", "qeema", "nihari"]

menu = []
menu_price = {}
menu_price_per_head = {}

guests = 300

while len(menu) < 5:
    dish = input("Enter dish name: ")

    if dish in discard_menu:
        print("This dish is not allowed!")
        continue

    menu.append(dish)

    price = int(input("Enter total price for 300 guests: "))
    menu_price[dish] = price

    per_head = round(price / guests, 2)
    menu_price_per_head[dish] = per_head

total_menu_price = sum(menu_price.values())
total_per_head_cost = round(sum(menu_price_per_head.values()), 2)

print("\nMenu:", menu)
print("Menu Prices:", menu_price)
print("Total Menu Price:", total_menu_price)
print("Per Head Cost of Each Dish:", menu_price_per_head)
print("Complete Menu Per Head Cost:", total_per_head_cost)
```
